```python

types = pd.json_normalize(data['types'])
stats = pd.json_normalize(data['stats'])
species = pd.json_normalize(data['species'])

cols_stats = ['hp','special-attack','attack','defense','special-defense','speed']

df = pd.concat(
    [
        species[['name']],
        types.reindex(columns=['type1','type2']),
        stats.reindex(columns=cols_stats),
    ],
    axis=1
)[['name','type1','type2'] + cols_stats]

df

```
