### 1.2  Base de datos unificada

Ahora junte los tres DataFrames en un nuevo DataFrame llamado `penguins`, agregando una nueva columna `species` que identifique el nombre de la especie de cada pingüino (Adelie, Chinstrap o Gentoo).

### 1.3 Inspección y limpieza

Inspeccione el DataFrame `penguins` y sus columnas. ¿Están correctamente asignados los tipos de cada columna? ¿Alguna podría ser categórica? ¿Existen valores nulos o inválidos? Aplique las transformaciones y limpiezas que considere necesarisas al DataFrame, explicando en los comentarios (o nuevas celdas) por qué hizo dichas transformaciones.

```python
import pandas as pd
import numpy as np

adelie    = pd.read_csv("penguins/adelie.csv")
chinstrap = pd.read_csv("penguins/chinstrap.csv")
gentoo    = pd.read_csv("penguins/gentoo.csv")

adelie["species"]    = "Adelie"
chinstrap["species"] = "Chinstrap"
gentoo["species"]    = "Gentoo"

pinwino = pd.concat([adelie, chinstrap, gentoo], ignore_index=True)
pinwino
```

```python
# convertir a numérico
columnas_numericas = ["culmen_length_mm", "culmen_depth_mm", "flipper_length_mm", "body_mass_g"]
for c in columnas_numericas:
    pinwino[c] = pd.to_numeric(pinwino[c], errors="coerce")

# estandarizar sex
pinwino["sex"] = pinwino["sex"].astype(str).str.strip().replace({"": pd.NA, "nan": pd.NA}).str.capitalize()
pinwino["sex"] = pinwino["sex"].astype("category")

# categoricas
pinwino["species"] = pinwino["species"].astype("category")
pinwino["island"]  = pinwino["island"].astype("category")

# dataset limpio y guardar
columnas_clave = ["culmen_length_mm", "culmen_depth_mm", "flipper_length_mm", "body_mass_g"]
pinwino_clean = pinwino.dropna(subset=columnas_clave, how="any").copy()
pinwino_clean.to_csv("pinwino_clean.csv", index=False)
```
