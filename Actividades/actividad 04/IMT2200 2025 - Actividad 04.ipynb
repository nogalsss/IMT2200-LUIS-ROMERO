```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
```

### 1. Extraer datos

#### 1.1 Respuesta:

```python
#type:ignore
url = "https://en.wikipedia.org/wiki/List_of_largest_cities"
html = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=30).text
soup = BeautifulSoup(html, "html.parser")

tabla = soup.find(id="List").find_next("table", class_="wikitable")
for sup in tabla.find_all("sup"):
    sup.decompose()

registros = []
for tr in tabla.find_all("tr")[1:]:
    celdas = [c.get_text(strip=True) for c in tr.find_all(["th", "td"])]
    if len(celdas) >= 3 and celdas[0].lower() != "definition":
        registros.append({"ciudad": celdas[0], "pais": celdas[1], "poblacion": celdas[2]})

df_ciudades = pd.DataFrame(registros)
df_ciudades.head(20)
```

#### 1.2 Respuesta:

```python
# 1.2
df_ciudades["poblacion"] = (
    df_ciudades["poblacion"]
      .str.split("(").str[0]      #  parentesis
      .str.replace(",", "")       #  comas
      .astype(int)
)

df_20_grandes = (
    df_ciudades
      .sort_values("poblacion", ascending=False)
      .head(20)
      .reset_index(drop=True)
)

df_20_grandes
```

### 2. Uso de API

#### 2.1 Respuesta:

```python
# 2.1

df_geo = df_20_grandes.copy()

latitudes, longitudes = [], []
i = 0
while i < len(df_geo):
    fila = df_geo.iloc[i]
    resp = requests.get(
        "https://geocoding-api.open-meteo.com/v1/search",
        params={"name": fila["ciudad"], "count": 10, "language": "en", "format": "json"},
        timeout=10
    ).json().get("results", [])

    if resp:
        elegido = next(
            (d for d in resp if d.get("country", "").lower() == fila["pais"].lower()),
            resp[0]
        )
        latitudes.append(elegido["latitude"])
        longitudes.append(elegido["longitude"])
    else:
        latitudes.append(np.nan)
        longitudes.append(np.nan)
    i += 1

df_geo["latitud"]  = latitudes
df_geo["longitud"] = longitudes
df_geo.head(20)
```
