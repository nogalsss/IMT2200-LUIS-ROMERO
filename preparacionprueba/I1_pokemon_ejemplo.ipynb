## Pregunta 1 (15 pts)
**Carga de datos y exploración básica**

1. Descarga (o abre localmente) el archivo `all_pokemon.json` dentro del ZIP y cárgalo en un `DataFrame` llamado `df`.
2. Muestra el número total de Pokémon y el número de columnas.
3. Presenta una muestra aleatoria de 5 filas.

```python
import zipfile, json, pandas as pd

# === Cargar JSON desde el ZIP ===
zip_path = 'all_pokemon.zip'  # ← cambia si es necesario
with zipfile.ZipFile(zip_path) as z:
    with z.open('all_pokemon.json') as f:
        data_raw = json.load(f)

df = pd.DataFrame(data_raw.values())
n_filas, n_cols = df.shape
print(f"Pokémon: {n_filas}  |  Columnas: {n_cols}")

df.sample(5)
```

## Pregunta 2 (15 pts)
**Limpieza y creación de variables**

1. Elimina cualquier Pokémon duplicado basado en la columna `id`.
2. Convierte `height` y `weight` a metros y kilogramos, guardando los resultados en `altura_m` y `peso_kg`.
3. Calcula el **Índice Corporal Pokémon** (IC = peso / altura²) y guárdalo en `indice_corporeo`.

```python
# 1) Duplicados
df = df.drop_duplicates(subset='id').reset_index(drop=True)

# 2) Unidades
df['altura_m'] = df['height'] / 10   # decímetros → metros
df['peso_kg']  = df['weight'] / 10   # hectogramos → kg

# 3) Índice corporal
df['indice_corporeo'] = df['peso_kg'] / df['altura_m']**2

df[['name', 'altura_m', 'peso_kg', 'indice_corporeo']].head()
```

## Pregunta 3 (10 pts)
**Top‑5 Pokémon más pesados**

1. Lista los cinco Pokémon con mayor peso (`peso_kg`) junto con su `primary_type`.
2. Comenta brevemente si observas predominio de algún tipo.

```python
# Obtener primary_type (primer elemento de la lista de tipos)
df['primary_type'] = df['types'].apply(lambda lst: lst[0]['type']['name'])

top5_pesados = (
    df[['name', 'primary_type', 'peso_kg']]
      .nlargest(5, 'peso_kg')
      .reset_index(drop=True)
)
top5_pesados
```

## Pregunta 4 (20 pts)
**Índice de poder y análisis por tipo**

1. Aplana el campo `stats` y crea un índice de poder `power_index` definido como la suma de los seis stats básicos.
2. Calcula el promedio de `power_index` por `primary_type` y muestra los tres tipos con mayor promedio.

```python
# 1) Aplanar stats
stats_cols = df['stats'].apply(pd.Series)[['hp', 'attack', 'defense', 'special-attack', 'special-defense', 'speed']]
stats_cols = stats_cols.rename(columns={
    'special-attack': 'sp_attack',
    'special-defense': 'sp_defense'
})
df = pd.concat([df.drop(columns=['stats']), stats_cols], axis=1)

df['power_index'] = df[['hp', 'attack', 'defense', 'sp_attack', 'sp_defense', 'speed']].sum(axis=1)

# 2) Promedio por tipo
tipo_power = (
    df.groupby('primary_type')['power_index']
      .mean()
      .sort_values(ascending=False)
      .head(3)
)
tipo_power
```

## Pregunta 5 (10 pts)
**Visualización**

Dibuja un diagrama de dispersión de `altura_m` vs `peso_kg` y colorea los puntos según `primary_type`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))
sns.scatterplot(
    data=df,
    x='altura_m', y='peso_kg',
    hue='primary_type',
    palette='tab20', legend=False
)
plt.xlabel('Altura (m)')
plt.ylabel('Peso (kg)')
plt.title('Altura vs Peso de Pokémon')
plt.tight_layout()
plt.show()
```
